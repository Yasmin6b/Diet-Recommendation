```python
import pandas as pd
import random
```

```python
# Define exercise categories
exercises = [
    ('Hip Thrust', 'Strengthen glutes with this powerful hip extension exercise.'),
    ('Lunges', 'A great lower-body exercise that targets quads, hamstrings, and glutes.'),
    ('Reverse Lunges', 'Step back instead of forward to work balance and stability.'),
    ('Single Leg Hip Thrust', 'Unilateral version of hip thrust focusing on one side at a time.'),
    ('Bicep Curls', 'Isolate and strengthen the biceps using controlled curling motion.'),
    ('Deadlifts', 'Full-body compound lift engaging hamstrings, glutes, and lower back.'),
    ('Squats', 'A fundamental lower-body exercise that builds strength in legs and core.'),
    ('Leg Press', 'Machine-based lower-body movement to build quad and glute strength.'),
    ('Calf Raises', 'Strengthens and tones the calves through upward motion.'),
    ('Pull-Ups', 'Upper-body compound movement targeting back and biceps.'),
    ('Bench Press', 'Chest-focused pressing movement for strength and hypertrophy.'),
    ('Shoulder Press', 'Overhead pressing movement engaging shoulders and triceps.'),
    ('Dips', 'Bodyweight exercise for triceps and chest strength.'),
    ('Plank', 'Core stability exercise that engages abs and lower back muscles.'),
    ('Russian Twists', 'Rotational core movement to strengthen obliques.'),
    ('Lat Pulldown', 'Machine-based exercise targeting the upper back and lats.'),
    ('Cable Rows', 'Seated pulling movement strengthening the back and arms.'),
    ('Hamstring Curls', 'Isolation exercise for hamstrings using a machine.'),
    ('Leg Extensions', 'Machine-based exercise to strengthen the quadriceps.'),
    ('Glute Bridges', 'Great for activating and strengthening the glutes.'),
    ('Kettlebell Swings', 'Full-body explosive movement focusing on posterior chain.'),
    ('Tricep Extensions', 'Isolation exercise for triceps using a dumbbell or cable.'),
    ('Arnold Press', 'Shoulder pressing movement with a rotation for full engagement.'),
    ('Hammer Curls', 'Variation of bicep curls targeting the brachialis muscle.'),
    ('Face Pulls', 'Great for shoulder health and rear delt activation.'),
    ('Step-Ups', 'Leg exercise improving single-leg strength and coordination.'),
    ('Hanging Leg Raises', 'Advanced core exercise focusing on lower abs.'),
    ('Mountain Climbers', 'Cardio-intensive core exercise improving endurance and agility.'),
    ('Jump Squats', 'Explosive squat variation for power and athleticism.')
]
```

```python
body_parts = ['Chest', 'Back', 'Quads', 'Hamstrings', 'Glutes', 'Calves', 'Shoulders', 'Arms', 'Core', 'Full Body']
equipments = ['Dumbbell', 'Barbell', 'Machine', 'Bodyweight', 'Kettlebell', 'Resistance Bands']
levels = ['Beginner', 'Intermediate', 'Advanced']
types = ['Strength', 'Hypertrophy', 'Endurance', 'Mobility']
```

```python

# Generate dataset with 200 exercises
data = []
for i in range(200):
    name, desc = random.choice(exercises)
    exercise = {
        'Title': name,
        'Desc': desc,
        'Type': random.choice(types),
        'BodyPart': random.choice(body_parts),
        'Equipment': random.choice(equipments),
        'Level': random.choice(levels),
        'Rating': round(random.uniform(3.0, 5.0), 1),
        'RatingDesc': 'Good' if random.random() > 0.2 else 'Average',
        'Calories_Burned': round(random.uniform(50, 300), 1),
        'Image_URL': f'https://example.com/{name.replace(" ", "_").lower()}.jpg'
    }
    data.append(exercise)
```

```python
# Convert to DataFrame
dataset = pd.DataFrame(data)

# Save to CSV
dataset.to_csv('full_body_gym_dataset.csv', index=False)

print("Full-body gym dataset with 200 exercises generated successfully!")
```
